## 1. Bibliotecas
Importamos las herramientas para leer el archivo y explorar la tabla.

```python
import pandas as pd
from zipfile import ZipFile
from io import BytesIO
from google.colab import files
from IPython.display import display
import hashlib

pd.set_option("display.max_columns", None)
```

## 2. Carga
Selecciona únicamente el ZIP de NAS. Revisamos su contenido antes de leer el CSV.

```python
archivos_subidos = files.upload()
nombres_zip = [n for n in archivos_subidos if n.lower().endswith(".zip")]
if len(nombres_zip) != 1:
    raise ValueError("Sube un único ZIP de NAS.")
nombre_zip = nombres_zip[0]
contenido_zip = archivos_subidos[nombre_zip]
print("Archivo:", nombre_zip)
print("SHA-256:", hashlib.sha256(contenido_zip).hexdigest())
with ZipFile(BytesIO(contenido_zip)) as archivo_zip:
    for nombre in archivo_zip.namelist():
        print("-", nombre)
```

### Lectura del CSV
La primera fila contiene la cabecera. UTF-8-sig interpreta la codificación y su marca inicial. No modificamos nombres ni valores.

```python
with ZipFile(BytesIO(contenido_zip)) as archivo_zip:
    with archivo_zip.open("NAS-Specimen-Download.csv") as archivo_csv:
        df_original = pd.read_csv(archivo_csv, encoding="utf-8-sig", low_memory=False)
display(df_original.head())
```

### Cita y aviso del proveedor
Mostramos la documentación incluida para mantener la procedencia junto al análisis.

```python
with ZipFile(BytesIO(contenido_zip)) as archivo_zip:
    for nombre in ["citation.txt", "USGS NAS Data Disclaimer.txt"]:
        if nombre in archivo_zip.namelist():
            print(nombre)
            print(archivo_zip.read(nombre).decode("utf-8-sig"))
```

## 3. Diagnóstico inicial
Contamos registros y columnas sin aplicar ninguna limpieza.

```python
print("Registros:", df_original.shape[0])
print("Columnas:", df_original.shape[1])
```

### Tipos, nulos y valores distintos
Los tipos inferidos por Pandas no determinan el significado de las variables: un identificador numérico no es una medida.

```python
resumen_columnas = pd.DataFrame({
    "tipo": df_original.dtypes.astype(str),
    "no_nulos": df_original.notna().sum(),
    "nulos": df_original.isna().sum(),
    "porcentaje_nulos": (df_original.isna().mean() * 100).round(2),
    "valores_distintos": df_original.nunique(dropna=True)
})
with pd.option_context("display.max_rows", None):
    display(resumen_columnas)
```

### Duplicados y vacíos
Detectamos incidencias; todavía no eliminamos filas ni columnas.

```python
print("Duplicados exactos:", df_original.duplicated().sum())
print("Filas completamente vacías:", df_original.isna().all(axis=1).sum())
columnas_vacias = df_original.columns[df_original.isna().all()].tolist()
print("Columnas completamente vacías:", len(columnas_vacias))
print(columnas_vacias)
```

### Identificador
Comprobamos Specimen Number. Identificadores únicos no descartan registros de la misma observación con números diferentes.

```python
identificador = df_original["Specimen Number"]
print("Identificadores nulos:", identificador.isna().sum())
print("Identificadores distintos:", identificador.nunique())
repetidos = identificador.notna() & identificador.duplicated(keep=False)
print("Filas con identificador repetido:", repetidos.sum())
display(df_original.loc[repetidos])
```

### Muestra reproducible
Revisamos registros de distintas partes de la tabla.

```python
display(df_original.sample(n=min(5, len(df_original)), random_state=8))
display(df_original.tail())
```

## 4. Reconocimiento de variables
Contamos categorías originales y ausentes. Antes de usarlas como filtros interpretaremos su significado en los metadatos.

```python
for columna in ["Scientific Name", "Accuracy", "Source", "Status", "record_type"]:
    print("\n", columna)
    display(df_original[columna].value_counts(dropna=False).rename("registros").to_frame())
```

### Cobertura preliminar
Los recuentos describen esta descarga, no toda la evidencia mundial. Todavía no hemos validado fechas completas ni precisión geográfica.

```python
print("Año mínimo:", df_original["Year"].min())
print("Año máximo:", df_original["Year"].max())
print("Países/territorios distintos:", df_original["Country"].nunique())
display(df_original[["Year", "Month", "Day", "Country", "Latitude", "Longitude"]].isna().sum().rename("nulos").to_frame())
display(df_original["Country"].value_counts(dropna=False).rename("registros").to_frame())
```
